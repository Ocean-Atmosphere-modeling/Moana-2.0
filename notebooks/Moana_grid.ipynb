```python
# import all the necessary files 
import xarray as xr 
import numpy as np 
import xroms 
import matplotlib.pyplot as plt 
```

```python
import cmocean as cmo
```

```python
# importing the grid file
data_folder = '/projects/f_omg_1/moa/moana'
file_name = 'grd_moana5km.nc'
ds = xr.open_dataset(data_folder + '/' + file_name)
ds['Vtransform'] =2
ds['Vstretching'] = 5
ds["zeta"] = xr.zeros_like(ds.h)
ds_grid, xgrid = xroms.roms_dataset(ds, include_3D_metrics=False)
ds_grid.xroms.set_grid(xgrid)
```

```python
ds_grid.h.cf.plot(cmap=cmo.cm.thermal)
```
