```python
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from pathlib import Path
from matplotlib.animation import FuncAnimation, PillowWriter
import matplotlib.ticker as mticker
```

```python
fig_dir = Path("/home/nn411/Projects/OceanAtm/Moana-2.0/Figures")
fig_dir.mkdir(exist_ok=True)

LAND = "#d9d4c7"                                   # <-- was undefined

lon = ds.lon_rho
lat = ds.lat_rho
sss = ds.salt.isel(s_rho=-1)

vmin, vmax = 34, 35.6                              # <-- NZ surface salinity range
day_times = pd.to_datetime(sss["ocean_time"].values)

fig = plt.figure(figsize=(10, 7))
ax = plt.axes(projection=ccrs.PlateCarree(central_longitude=180))   # <-- domain crosses 180°
ax.set_extent([float(lon.min()), float(lon.max()), float(lat.min()), float(lat.max())],
              crs=ccrs.PlateCarree())

mesh = ax.pcolormesh(lon, lat, sss.isel(ocean_time=0),   # <-- the data was missing
                     cmap="YlGnBu_r", vmin=vmin, vmax=vmax,
                     transform=ccrs.PlateCarree(), shading="auto")
ax.add_feature(cfeature.LAND, facecolor=LAND, zorder=2)
ax.coastlines(resolution="10m", linewidth=0.5, zorder=3)
gl = ax.gridlines(crs=ccrs.PlateCarree(), draw_labels=True,
                  linestyle="--", color="gray", alpha=0.4)
gl.top_labels = False
gl.right_labels = False
gl.xlocator = mticker.FixedLocator([165, 170, 175, 180, -175])   # <-- longitudes to label
gl.ylocator = mticker.FixedLocator([-50, -45, -40, -35])         # <-- latitudes to label
gl.xlabel_style = {"color": "black"}
gl.ylabel_style = {"color": "black"}
gl.top_labels = False
gl.right_labels = False
gl.xlabel_style = {"color": "black"}
gl.ylabel_style = {"color": "black"}

cb = fig.colorbar(mesh, ax=ax, shrink=0.8, pad=0.03, extend="both")
cb.ax.tick_params(colors="black")
cb.set_label("Sea surface salinity (PSU)", color="black")
date_text = ax.text(0.02, 0.97, "", transform=ax.transAxes, va="top",
                    fontsize=15, fontweight="bold", color="black", zorder=10,
                    bbox=dict(facecolor="white", alpha=0.85, edgecolor="none"))
ax.set_title("ROMS Surface Salinity, Jan 2026", fontsize=15, color="black")   # <--

def update(i):
    mesh.set_array(sss.isel(ocean_time=i).values)
    date_text.set_text(day_times[i].strftime("%d %b %Y %H:%M"))   # hour added
    return mesh, date_text

# --- Test one frame first ---
update(0)                                          # <-- first day of the run
fig.savefig(fig_dir / "roms_sss_test.png", dpi=100)

anim = FuncAnimation(fig, update, frames=len(day_times))
anim.save(fig_dir / "roms_sss_week.gif", writer=PillowWriter(fps=12), dpi=100)
```

```python
# --- Surface relative vorticity on rho points ---
u = ds.u_eastward.isel(s_rho=-1).values    # (time, eta_rho, xi_rho)
v = ds.v_northward.isel(s_rho=-1).values
pm = ds.pm.values                          # 1/dx
pn = ds.pn.values                          # 1/dy
f  = ds.f.values

dvdx = np.gradient(v / pn, axis=-1)        # change of v·dy across one cell in x
dudy = np.gradient(u / pm, axis=-2)        # change of u·dx across one cell in y

zeta = pm * pn * (dvdx - dudy)             # 1/s
ro = zeta / f                              # ζ/f
ro = np.where(ds.mask_rho.values == 1, ro, np.nan)   # hide land

print(ro.shape, np.nanpercentile(ro, [1, 99]))
```

```python
fig_dir = Path("/home/nn411/Projects/OceanAtm/Moana-2.0/Figures")
fig_dir.mkdir(exist_ok=True)

LAND = "#d9d4c7"

lon = ds.lon_rho                          # <-- rho grid, same as ro
lat = ds.lat_rho
vmin, vmax = -0.3, 0.3                    # <-- matches your 1st–99th percentile

day_times = pd.to_datetime(ds["ocean_time"].values)

fig = plt.figure(figsize=(10, 7))
ax = plt.axes(projection=ccrs.PlateCarree(central_longitude=180))
ax.set_extent([float(lon.min()), float(lon.max()), float(lat.min()), float(lat.max())],
              crs=ccrs.PlateCarree())

mesh = ax.pcolormesh(lon, lat, ro[0], cmap="RdBu_r", vmin=vmin, vmax=vmax,
                     transform=ccrs.PlateCarree(), shading="nearest")
ax.add_feature(cfeature.LAND, facecolor=LAND, zorder=2)
ax.coastlines(resolution="10m", linewidth=0.5, zorder=3)

gl = ax.gridlines(crs=ccrs.PlateCarree(), draw_labels=True,
                  linestyle="--", color="gray", alpha=0.4)
gl.top_labels = False
gl.right_labels = False
gl.xlocator = mticker.FixedLocator([165, 170, 175, 180, -175])
gl.ylocator = mticker.FixedLocator([-50, -45, -40, -35])
gl.xlabel_style = {"color": "black"}
gl.ylabel_style = {"color": "black"}

cb = fig.colorbar(mesh, ax=ax, shrink=0.8, pad=0.03, extend="both")   # <-- restored
cb.ax.tick_params(colors="black")
cb.set_label("Surface relative vorticity ζ/f", color="black")

date_text = ax.text(0.02, 0.97, "", transform=ax.transAxes, va="top",
                    fontsize=15, fontweight="bold", color="black", zorder=10,
                    bbox=dict(facecolor="white", alpha=0.85, edgecolor="none"))
ax.set_title("ROMS Surface Vorticity, Jan 2026", fontsize=15, color="black")

def update(i):
    mesh.set_array(ro[i])
    date_text.set_text(day_times[i].strftime("%d %b %Y %H:%M"))
    return mesh, date_text

# --- Test one frame first ---
update(0)
fig.savefig(fig_dir / "roms_vort_test.png", dpi=100)          # <-- new name

anim = FuncAnimation(fig, update, frames=len(day_times))
anim.save(fig_dir / "roms_vort_week.gif", writer=PillowWriter(fps=12), dpi=100)
```
